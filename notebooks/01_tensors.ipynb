By the end of this notebook, you will be able to:

```python
import torch
print("PyTorch version", torch.__version__)
print("CUDA available", torch.cuda.is_available())
```

## 1. Creating tensors

Tensors are the core data structure in PyTorch.
Yor can create them from:
- Python lists
- NumPy arrays
- Factory methods like 'zeros', 'ones', 'rand', 'arrange'

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([[1, 2, 3], [4, 5, 6]])
a, b
```

```python
print("a:", a)
print("shape:", a.shape)
print("dtype:", a.dtype)
print("device:", a.device)
```

## 2. Tensor dtypes
`dtype` controls the type of numbers in a tensor (int, float, etc).
This matters for:
- memory usuage
- model training (most training uses float32)

```python
x_int = torch.tensor([1, 2, 3], dtype=torch.int)
x_float = torch.tensor([1, 2, 3], dtype=torch.float)
print(x_int, x_int.dtype)
print(x_float, x_float.dtype)
```

```python
x = torch.tensor([1, 2, 3])
x = x.to(torch.float32)
print(x, x.dtype)
```

## 3. Factory methods

```python
z = torch.zeros((2, 3))
o = torch.ones((2, 3))
r = torch.rand((2, 3))         # random in [0, 1]
seq = torch.arange(0, 10, 2)   # 0,2,4,6,8
z, o, r, seq
```

```python
like_z = torch.zeros_like(r)
like_r = torch.rand_like(o)
like_z, like_r
```

## 4. Shape and reshaping

Shpaes are critical in deep learning.

Common reshape operations:
- `reshape`
- `view` (similar but needs contiguous memory)
- `squeeze` / `unsqueeze`

```python
t = torch.arange(0, 12) # 12 elements
t2 = t.reshape(3, 4)
print("t shape:", t.shape)
print("t2 shape:", t2.shape)
t2
```

```python
v = torch.tensor([10, 20, 30]) # shape (3, )

v_unsq0 = v.unsqueeze(0) # add dimension at front -> (1, 3)
v_unsq1 = v.unsqueeze(1) # add dimension at end -> (3, 1)

print("v:", v.shape)
print("unsqueeze(0):", v_unsq0.shape)
print("unsqueeze(1):", v_unsq1.shape)

print("squeeze back:", v_unsq0.squeeze(0).shape)
```

## 5. Indexing and slicing

Indexing tesnsors is similar to NumPy.

```python
m = torch.tensor([[10, 20, 30],
                  [40, 50, 60]])
m.shape
```

## 6. Basic tensor operations

PyTorch supports:
- elementswise ops: +, -, *, /
- matrix multipication: `@` or `torch.matmul`
- reductions: sum, mean, max

```python
p = torch.tensor([1.0, 2.0, 3.0])
q = torch.tensor([10.0, 20.0, 30.0])

print("p + q:", p + q)
print("p - q:", p - q)
print("p * q:", p * q)
print("p / q:", p / q)
```

```python
A = torch.rand((2, 3))
B = torch.rand((3, 4))

C = A @ B
print("A shape:", A.shape)
print("B shape:", B.shape)
print("C shape:", C.shape)
```

```python
x = torch.tensor([[1.0, 2.0],
                  [3.0, 4.0]])
print("x sum:", x.sum())
print("x mean:", x.mean())
print("x max:", x.max())
print("argmax", x.argmax())
```

## 7. Broadcasting
Broadcasting lets PyTorch combine tensors of different shapes by automatically expanding dimensions when possible.

This is extremely common in deep learning.

```python
M = torch.ones((3,4))
v = torch.tensor([1.0, 2.0, 3.0, 4.0]) # shape (4, )

out = M + v # v broadcasts across rows
print(M)
print(v)
print("M shape:", M.shape)
print("v shape:", v.shape)
print("out shape:", out.shape)
out
```

```python
# Example: adding a column vector
col = torch.tensor([10.0, 20.0, 30.0]).unsqueeze(1) # shape (3, 1)
print(col)
out2 = M + col
print("col shape", col.shape)
print("out2 shape", out2.shape)
out2
```

## 8. CPU vs GPU (device)

If CUDA is available, you can move tensors to GPU for faster compute.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using devide:", device)

x = torch.rand((2, 2))
x = x.to(device)
print("x device:", x.device)
```

## Common mistakes
- Forgetting to check shapes (`tensor.shape`)
- Mixing dtypes (int vs float) when computing
- Mixing devices (CPU tensor + GPU tensor) causes errors
- Using `@` when you meant elementwise `*` (and vice versa)
