```python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

## What is PyTorch?

PyTorch is a deep learning framework that provides:
- **Tensors** (multi-dimensional arrays) with GPU acceleration
- **Automatic differentiation** (autograd)
- A flexible, Python-first API for building neural networks

PyTorch is widely used in:
- Research
- Production machine learning
- Deep learning systems


```python
# Creating a tensor
x = torch.tensor([1, 2, 3])
x
```

```python
x.shape, x.dtype, x.device
```

## PyTorch vs NumPy

PyTorch tensors are similar to NumPy arrays, but with key advantages:
- GPU acceleration
- Automatic differentiation
- Integration with deep learning workflows

```python
import numpy as np
np_array = np.array([1, 2, 3])
torch_tensor = torch.tensor([1, 2, 3])
type(np_array), type(torch_tensor)
```

## Common beginner mistakes
- Treating tensors like plain Python lists
- Ignoring tensor shapes
- Mixing CPU and GPU tensors
